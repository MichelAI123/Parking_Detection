```python
import os
import cv2
import easyocr
import matplotlib.pyplot as plt
from ultralytics import YOLO

# --- DÉFINITION DYNAMIQUE DU CHEMIN (STANDARD DOCKER) ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

# --- 1. CHARGEMENT DES MODÈLES (Une seule fois au démarrage) ---
chemin_modele = os.path.join(BASE_DIR, "best.pt")

if not os.path.exists(chemin_modele):
    print(f"Alerte : Modèle non trouvé au chemin {chemin_modele}. Utilisation du modèle de base.")
    chemin_modele = "yolov8n.pt"

print("Initialisation des modèles (YOLO + EasyOCR)...")
model = YOLO(chemin_modele)
reader = easyocr.Reader(['en'])

# Liste des textes à ignorer (provinces et slogans)
MOTS_IGNORER = ["ONTARIO", "QUEBEC", "YOURSTODISCOVER", "JEMESOUVIENS", "CANADA"]

def tester_ocr_sur_image(nom_image):
    """
    Localise la plaque avec YOLO, la découpe proprement, et lit le texte avec EasyOCR.
    """
    # Résolution dynamique du chemin de l'image
    chemin_image = os.path.join(BASE_DIR, "images", nom_image)
    
    img = cv2.imread(chemin_image)
    if img is None:
        print(f"Erreur critique : Image introuvable au chemin {chemin_image}")
        return

    # Étape 1 : Localisation avec YOLO (en mémoire)
    results = model.predict(source=img, conf=0.5, verbose=False)
    
    plaque_trouvee = False

    for r in results:
        for box in r.boxes:
            plaque_trouvee = True
            
            # Récupérer les coordonnées du rectangle
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            
            # Sécurité d'ingénierie : Prévention des dépassements d'index (Out of Bounds)
            y1, y2 = max(0, y1), min(img.shape[0], y2)
            x1, x2 = max(0, x1), min(img.shape[1], x2)
            
            # Étape 2 : Découpage (Crop) de la plaque
            plaque_crop = img[y1:y2, x1:x2]
            
            # Étape 3 : Lecture avec EasyOCR
            resultat_lecture = reader.readtext(plaque_crop)
            
            plaque_finale = ""
            max_conf = 0.0
            
            for (bbox, text, prob) in resultat_lecture:
                # Nettoyage : Majuscules, retrait des espaces et tirets
                text_clean = text.upper().replace(" ", "").replace("-", "").replace(".", "")
                
                # Étape 4 : Filtrage intelligent
                if prob > 0.4 and len(text_clean) >= 4 and text_clean not in MOTS_IGNORER:
                    if prob > max_conf:
                        max_conf = prob
                        plaque_finale = text_clean
            
            # --- AFFICHAGE DES RÉSULTATS ---
            plt.figure(figsize=(6, 3))
            img_display = cv2.cvtColor(plaque_crop, cv2.COLOR_BGR2RGB)
            plt.imshow(img_display)
            
            if plaque_finale:
                print(f"NUMÉRO DÉTECTÉ : {plaque_finale} (Confiance: {max_conf:.2f})")
                plt.title(f"Plaque : {plaque_finale}")
            else:
                print("Alerte : Plaque localisée mais aucun numéro valide n'a pu être lu.")
                plt.title("Lecture échouée / filtrée")
                
            plt.axis('off')
            plt.show()

    if not plaque_trouvee:
        print("Aucun véhicule ou plaque détecté par YOLO sur cette image.")

# --- BLOC D'EXÉCUTION PRINCIPAL ---
if __name__ == "__main__":
    # Test avec l'image 7.jpg située dans le dossier images
    tester_ocr_sur_image("7.jpg")
```

```python
import os
import cv2
import easyocr
import matplotlib.pyplot as plt
from ultralytics import YOLO

# --- DÉFINITION DYNAMIQUE DU CHEMIN (STANDARD DOCKER) ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

# --- 1. CHARGEMENT DES MODÈLES (Au démarrage du serveur) ---
chemin_modele = os.path.join(BASE_DIR, "best.pt")

if not os.path.exists(chemin_modele):
    print(f"Alerte : Modèle non trouvé au chemin {chemin_modele}. Utilisation du modèle par défaut.")
    chemin_modele = "yolov8n.pt"

print("Initialisation des modèles (YOLO + EasyOCR)...")
model = YOLO(chemin_modele)
reader = easyocr.Reader(['en'])

MOTS_IGNORER = ["ONTARIO", "QUEBEC", "YOURSTODISCOVER", "JEMESOUVIENS", "CANADA", "ONT", "QC"]

def extraire_texte_plaque(nom_image):
    """
    Détecte la plaque, découpe l'image de manière sécurisée et extrait le texte filtré.
    """
    chemin_image = os.path.join(BASE_DIR, "images", nom_image)
    
    img = cv2.imread(chemin_image)
    if img is None:
        print(f"Erreur critique : Impossible de lire l'image à l'emplacement {chemin_image}")
        return

    hauteur_img, largeur_img = img.shape[:2]

    # Étape 1 : Détection avec YOLO directement depuis la RAM
    results = model.predict(source=img, conf=0.5, verbose=False)
    
    found_any = False

    for r in results:
        for box in r.boxes:
            found_any = True
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            
            # Étape 2 : Marge de 5 pixels AVEC protection contre les dépassements d'index
            y1_crop = max(0, y1 - 5)
            y2_crop = min(hauteur_img, y2 + 5)
            x1_crop = max(0, x1 - 5)
            x2_crop = min(largeur_img, x2 + 5)
            
            plaque_crop = img[y1_crop:y2_crop, x1_crop:x2_crop]
            
            # Étape 3 : Lecture du texte par EasyOCR
            resultats_ocr = reader.readtext(plaque_crop)
            
            plaque_finale = ""
            meilleure_confiance = 0.0
            
            # Étape 4 : Analyse et Filtrage
            for (bbox, text, prob) in resultats_ocr:
                text_propre = text.upper().replace(" ", "").replace("-", "").replace(".", "")
                
                if prob > 0.3 and len(text_propre) >= 4 and text_propre not in MOTS_IGNORER:
                    if prob > meilleure_confiance:
                        meilleure_confiance = prob
                        plaque_finale = text_propre
            
            # --- AFFICHAGE DES RÉSULTATS ---
            plt.figure(figsize=(6, 3))
            img_rgb = cv2.cvtColor(plaque_crop, cv2.COLOR_BGR2RGB)
            plt.imshow(img_rgb)
            
            if plaque_finale:
                print(f" NUMÉRO DÉTECTÉ : {plaque_finale} (Score: {meilleure_confiance:.2f})")
                plt.title(f"Lecture OCR : {plaque_finale}")
            else:
                print(" Plaque localisée, mais le texte est illisible ou filtré.")
                plt.title("Lecture OCR : Échec/Filtré")
                
            plt.axis('off')
            plt.show()

    if not found_any:
        print(" Aucune plaque détectée sur cette image.")

# --- BLOC D'EXÉCUTION PRINCIPAL ---
# Ce bloc permet de tester le fichier indépendamment
if __name__ == "__main__":
    # Test avec l'image 7.jpg du dossier images
    extraire_texte_plaque("7.jpg")
```

```python
import os
import cv2
import easyocr

# --- DÉFINITION DYNAMIQUE DU CHEMIN (STANDARD LOGICIEL) ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

def tester_ocr_complet(nom_image):
    """
    Script de diagnostic : Analyse l'image complète avec EasyOCR sans pré-découpage YOLO.
    Permet d'évaluer le 'bruit' visuel (textes parasites).
    """
    # Pointage dynamique vers le dossier images
    chemin_image = os.path.join(BASE_DIR, "images", nom_image)

    if not os.path.exists(chemin_image):
        print(f"Erreur de diagnostic : L'image {nom_image} est introuvable au chemin : {chemin_image}")
        print("Vérifiez si l'extension est correcte (.jpg au lieu de .jpeg)")
        return

    print(f"Chargement de l'image de test : {chemin_image}")
    img = cv2.imread(chemin_image)
    
    if img is None:
        print("Erreur critique : Impossible de décoder l'image (fichier potentiellement corrompu).")
        return

    print("Initialisation du lecteur EasyOCR...")
    # L'argument gpu=True accélérera drastiquement cette étape si vous avez une carte graphique compatible
    reader = easyocr.Reader(['en'], gpu=True)

    print("Analyse de l'image entière en cours (Extraction du bruit visuel)...")
    resultats = reader.readtext(img)

    print("\n--- TOUS LES TEXTES DÉTECTÉS (SANS FILTRE YOLO) ---")
    if not resultats:
        print("Aucun texte détecté.")
    else:
        for (bbox, texte, conf) in resultats:
            print(f"  → '{texte}'  (confiance: {conf:.2f})")
    print("----------------------------------------------------")

# --- BLOC D'EXÉCUTION PRINCIPAL ---
if __name__ == "__main__":
    # Test ciblé sur votre fichier (modifiez en "89.jpg" si le terminal ne le trouve pas)
    tester_ocr_complet("89.jpeg")
```

```python
import os
import re
import cv2
import easyocr
import matplotlib.pyplot as plt
from ultralytics import YOLO

# --- DÉFINITION DYNAMIQUE DU CHEMIN (STANDARD LOGICIEL) ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

# --- 1. CONFIGURATION ET CHARGEMENT DES MODÈLES ---
chemin_modele = os.path.join(BASE_DIR, "best.pt")

if not os.path.exists(chemin_modele):
    print(f"Alerte : Modèle non trouvé au chemin {chemin_modele}. Utilisation du modèle de base.")
    chemin_modele = "yolov8n.pt"

print("Initialisation des modèles (YOLO + EasyOCR)...")
model = YOLO(chemin_modele)
reader = easyocr.Reader(['en'])

# Liste noire des mots à bannir totalement
LISTE_NOIRE = ["QUEBEC", "ONTARIO", "CANADA", "JEMESOUVIENS", "YOURSTODISCOVER"]

def detecter_et_lire_final(nom_image):
    """
    Pipeline avancé : Détection YOLO, marge dynamique 10%, Otsu et OCR filtré.
    """
    # Résolution dynamique du chemin (pointant vers le sous-dossier images/new)
    chemin_image = os.path.join(BASE_DIR, "images", "new", nom_image)
    
    img = cv2.imread(chemin_image)
    if img is None:
        print(f"Erreur critique : L'image {nom_image} est introuvable au chemin : {chemin_image}")
        return

    # 1. Localisation avec YOLO (en mémoire vive)
    results = model.predict(source=img, conf=0.5, verbose=False)
    
    plaque_trouvee = False

    for r in results:
        for box in r.boxes:
            plaque_trouvee = True
            
            # Récupérer les coordonnées brutes
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            
            # --- ASTUCE A : Marge dynamique sécurisée de 10% ---
            h, w = y2 - y1, x2 - x1
            y1 = max(0, y1 - int(h * 0.1))
            y2 = min(img.shape[0], y2 + int(h * 0.1))
            x1 = max(0, x1 - int(w * 0.1))
            x2 = min(img.shape[1], x2 + int(w * 0.1))
            
            plaque_crop = img[y1:y2, x1:x2]
            
            # --- ASTUCE B : Prétraitement (Gris + Otsu) ---
            gray = cv2.cvtColor(plaque_crop, cv2.COLOR_BGR2GRAY)
            plaque_prete = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]

            # 2. Lecture OCR
            result_ocr = reader.readtext(plaque_prete)
            
            candidats = []
            for (bbox, text, prob) in result_ocr:
                # NETTOYAGE : A-Z et 0-9 uniquement
                t_clean = re.sub(r'[^A-Z0-9]', '', text.upper())
                
                # Règles d'affaires strictes
                if t_clean not in LISTE_NOIRE and len(t_clean) >= 4:
                    if any(char.isdigit() for char in t_clean):
                        candidats.append((t_clean, prob))
            
            # Affichage des résultats
            plt.figure(figsize=(6, 3))
            if candidats:
                # Tri des candidats par indice de confiance décroissant
                candidats.sort(key=lambda x: x[1], reverse=True)
                plaque_finale, confiance = candidats[0]
                
                print(f"PLAQUE EXTRAITE : {plaque_finale} (Confiance : {confiance:.2f})")
                plt.imshow(plaque_prete, cmap='gray')
                plt.title(f"Zone traitée : {plaque_finale}")
            else:
                print("Échec de lecture ou texte filtré par les règles d'affaires.")
                plt.imshow(plaque_prete, cmap='gray')
                plt.title("Rejet OCR")
                
            plt.axis('off')
            plt.show()

    if not plaque_trouvee:
        print("Aucun véhicule ou plaque détecté par YOLO sur cette image.")

# --- BLOC D'EXÉCUTION PRINCIPAL ---
if __name__ == "__main__":
    # Test ciblé sur le dossier "new" 
    tester_nom_image = "IMG_4201.jpg"
    detecter_et_lire_final(tester_nom_image)
```

```python
import os
import re
import cv2
import easyocr
import matplotlib.pyplot as plt
from ultralytics import YOLO

# --- DÉFINITION DYNAMIQUE DU CHEMIN (STANDARD LOGICIEL) ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

# --- 1. CONFIGURATION ET CHARGEMENT ---
chemin_modele = os.path.join(BASE_DIR, "best.pt")

if not os.path.exists(chemin_modele):
    print(f"Alerte : Modèle non trouvé au chemin {chemin_modele}. Utilisation du modèle de base.")
    chemin_modele = "yolov8n.pt"

print("Initialisation des modèles (YOLO + EasyOCR)...")
model = YOLO(chemin_modele)
reader = easyocr.Reader(['en'])

LISTE_NOIRE = ["QUEBEC", "ONTARIO", "CANADA", "JEMESOUVIENS", "YOURSTODISCOVER"]

def ameliorer_image(img_crop):
    """
    Pipeline de prétraitement avancé pour maximiser l'extraction OCR.
    """
    # Conversion en niveaux de gris
    gray = cv2.cvtColor(img_crop, cv2.COLOR_BGR2GRAY)
    # Agrandissement (x2) pour que les lettres soient plus nettes
    gray = cv2.resize(gray, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
    # Filtre pour enlever le bruit tout en préservant les contours
    gray = cv2.bilateralFilter(gray, 11, 17, 17)
    # Binarisation (Noir et Blanc pur)
    _, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    return thresh

def detecter_et_lire_final(nom_image):
    """
    Pipeline complet : Inférence, découpage sécurisé, amélioration et OCR.
    """
    # Résolution dynamique ciblant le dossier images/new/
    chemin_image = os.path.join(BASE_DIR, "images", "new", nom_image)
    
    img = cv2.imread(chemin_image)
    if img is None:
        print(f"Erreur critique : L'image {nom_image} est introuvable au chemin : {chemin_image}")
        return

    # Inférence directement depuis la mémoire vive
    results = model.predict(source=img, conf=0.5, verbose=False)
    
    plaque_trouvee = False

    for r in results:
        for box in r.boxes:
            plaque_trouvee = True
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            
            # Marge de sécurité de 5 pixels (avec protection contre les débordements)
            y1, y2 = max(0, y1 - 5), min(img.shape[0], y2 + 5)
            x1, x2 = max(0, x1 - 5), min(img.shape[1], x2 + 5)
            
            plaque_crop = img[y1:y2, x1:x2]
            
            # --- PRÉTRAITEMENT MAGIQUE ---
            plaque_prete = ameliorer_image(plaque_crop)
            
            # Lecture OCR sur l'image binarisée et agrandie
            result_ocr = reader.readtext(plaque_prete)
            
            candidats = []
            for (bbox, text, prob) in result_ocr:
                t_clean = re.sub(r'[^A-Z0-9]', '', text.upper())
                
                # Filtrage métier strict
                if t_clean not in LISTE_NOIRE and len(t_clean) >= 4:
                    if any(char.isdigit() for char in t_clean):
                        candidats.append((t_clean, prob))
            
            # Affichage des résultats
            plt.figure(figsize=(6, 3))
            if candidats:
                # Tri pour garder la meilleure probabilité
                candidats.sort(key=lambda x: x[1], reverse=True)
                plaque_finale, confiance = candidats[0]
                
                print(f" PLAQUE DÉTECTÉE : {plaque_finale} (Confiance: {confiance:.2f})")
                plt.imshow(plaque_prete, cmap='gray')
                plt.title(f"Lecture : {plaque_finale}")
            else:
                print(" Échec de lecture ou texte rejeté par le filtre d'affaires.")
                plt.imshow(plaque_prete, cmap='gray')
                plt.title("Rejet OCR")
                
            plt.axis('off')
            plt.show()

    if not plaque_trouvee:
        print("Aucun véhicule ou plaque détecté par YOLO sur cette image.")

# --- BLOC D'EXÉCUTION PRINCIPAL ---
if __name__ == "__main__":
    # Testez ici avec l'image souhaitée de votre dossier 'new'
    detecter_et_lire_final("IMG_4214.jpg")
```

```python
import os
import cv2
import re
import easyocr
import mlflow
from ultralytics import YOLO

# --- DÉFINITION DYNAMIQUE DU CHEMIN (STANDARD LOGICIEL) ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

# --- 1. CONFIGURATION MLOPS SÉCURISÉE ---
chemin_mlruns = os.path.join(BASE_DIR, "mlruns")
mlflow.set_tracking_uri(f"file:///{chemin_mlruns.replace(os.sep, '/')}")
mlflow.set_experiment("Parking_Control_Video_Tests")

# --- 2. INITIALISATION DES MODÈLES ---
chemin_modele = os.path.join(BASE_DIR, "best.pt")
if not os.path.exists(chemin_modele):
    print(f"Alerte : Modèle non trouvé au chemin {chemin_modele}. Utilisation du modèle de base.")
    chemin_modele = "yolov8n.pt"

print("Chargement des modèles (YOLO + EasyOCR)...")
model = YOLO(chemin_modele)
reader = easyocr.Reader(['en'])

LISTE_NOIRE = ["QUEBEC", "ONTARIO", "CANADA"]

def traiter_flux_video(nom_video, mode_headless=False):
    """
    Analyse un flux vidéo. 
    mode_headless=True est OBLIGATOIRE si exécuté dans Docker ou via une API Streamlit/FastAPI.
    """
    chemin_video = os.path.join(BASE_DIR, "videos", nom_video)
    
    if not os.path.exists(chemin_video):
        print(f"Erreur critique : La vidéo {nom_video} est introuvable au chemin : {chemin_video}")
        return

    cap = cv2.VideoCapture(chemin_video)
    if not cap.isOpened():
        print("Erreur : Impossible d'ouvrir le flux vidéo.")
        return

    derniere_plaque_lue = ""
    compteur_frames = 0
    
    print(f"Début de l'analyse vidéo : {nom_video} (Headless: {mode_headless})")

    # 3. SESSION MLOPS POUR LA VIDÉO
    with mlflow.start_run(run_name=f"Analyse_{nom_video}"):
        mlflow.log_param("fichier_video", nom_video)
        mlflow.log_param("frame_skip", 5) # On note qu'on analyse 1 frame sur 5
        
        while cap.isOpened():
            ret, frame = cap.read()
            if not ret:
                break # Fin de la vidéo
            
            compteur_frames += 1
            
            # OPTIMISATION : On n'analyse qu'une image sur 5 pour garder de la fluidité
            if compteur_frames % 5 != 0:
                continue

            # Détection avec YOLO (en RAM)
            results = model.predict(source=frame, conf=0.6, verbose=False)
            
            for r in results:
                for box in r.boxes:
                    # Sécurisation des limites du découpage vidéo
                    x1, y1, x2, y2 = map(int, box.xyxy[0])
                    y1, y2 = max(0, y1), min(frame.shape[0], y2)
                    x1, x2 = max(0, x1), min(frame.shape[1], x2)
                    
                    plaque_crop = frame[y1:y2, x1:x2]
                    
                    # Ignorer les crops vides causés par des détections hors cadre
                    if plaque_crop.size == 0:
                        continue
                    
                    # Lecture OCR
                    result_ocr = reader.readtext(plaque_crop)
                    
                    for (_, text, prob) in result_ocr:
                        t_clean = re.sub(r'[^A-Z0-9]', '', text.upper())
                        
                        # Logique métier d'identification
                        if len(t_clean) >= 4 and any(c.isdigit() for c in t_clean):
                            if t_clean not in LISTE_NOIRE and t_clean != derniere_plaque_lue:
                                print(f" Véhicule détecté : {t_clean} (Confiance: {prob:.2f})")
                                derniere_plaque_lue = t_clean
                                
                                # Enregistrement dans MLflow
                                mlflow.log_text(t_clean, f"plates/frame_{compteur_frames}.txt")
                                mlflow.log_metric("confiance_detection", float(prob), step=compteur_frames)
                                
                                # Feedback visuel sur la trame
                                cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
                                cv2.putText(frame, t_clean, (x1, max(0, y1 - 10)), 
                                            cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

            # --- AFFICHAGE SELON L'ENVIRONNEMENT ---
            if not mode_headless:
                # Si on est sur Windows/PC local, on affiche la fenêtre
                cv2.imshow("Controle Parking La Cite - IA", frame)
                if cv2.waitKey(1) & 0xFF == ord('q'):
                    print("Interruption manuelle par l'utilisateur.")
                    break
            else:
                # Si on est sur Docker, on pourrait sauvegarder une image ponctuellement
                if compteur_frames % 50 == 0:
                    cv2.imwrite(os.path.join(BASE_DIR, f"snapshot_frame_{compteur_frames}.jpg"), frame)

    cap.release()
    if not mode_headless:
        cv2.destroyAllWindows()
    
    print("Analyse vidéo terminée et session MLflow fermée proprement.")

# --- BLOC D'EXÉCUTION PRINCIPAL ---
if __name__ == "__main__":
    # Test local : mettez mode_headless=False pour voir la vidéo
    # Pour Docker ou l'API, vous passerez mode_headless=True
    traiter_flux_video("IMG_4217.MOV", mode_headless=False)
```

```python
import os
import cv2
import re
import easyocr
import mlflow 
from ultralytics import YOLO
from collections import Counter

# --- DÉFINITION DYNAMIQUE DU CHEMIN ---
BASE_DIR = os.path.dirname(os.path.abspath(__file__))

BASE_ABONNES = ["DFHP360", "CYLR602", "DJAV098"]
LISTE_NOIRE_PROVINCES = ["QUEBEC", "ONTARIO", "CANADA", "YOURSTODISCOVER", "JEMESOUVIENS"]
SEUIL_CONFIANCE_OCR = 0.20

# --- 1. CONFIGURATION MLOPS SÉCURISÉE ---
chemin_mlruns = os.path.join(BASE_DIR, "mlruns")
mlflow.set_tracking_uri(f"file:///{chemin_mlruns.replace(os.sep, '/')}")
mlflow.set_experiment("Parking_Control_System")

# --- 2. CONFIGURATION & CHARGEMENT DES MODÈLES ---
chemin_modele = os.path.join(BASE_DIR, "best.pt")
if not os.path.exists(chemin_modele):
    chemin_modele = "yolov8n.pt"

print("Chargement des modèles (YOLO + EasyOCR)...")
model = YOLO(chemin_modele)
reader = easyocr.Reader(['en'])

def ameliorer_plaque(img_crop):
    """Améliore le contraste pour l'OCR"""
    gray = cv2.cvtColor(img_crop, cv2.COLOR_BGR2GRAY)
    gray = cv2.resize(gray, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)
    thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
    return thresh

def analyser_video_complete(nom_video, mode_headless=False):
    chemin_video = os.path.join(BASE_DIR, "videos", nom_video)
    
    if not os.path.exists(chemin_video):
        print(f"Erreur : Vidéo introuvable au chemin {chemin_video}")
        return

    cap = cv2.VideoCapture(chemin_video)
    memoire_plaques = []
    toutes_plaques_detectees = set() # Pour le résumé MLflow
    derniere_plaque_affichee = ""

    # DÉMARRAGE SÉCURISÉ DE LA SESSION MLFLOW
    with mlflow.start_run(run_name=f"Analyse_{nom_video.split('.')[0]}"):
        
        # --- ENREGISTREMENT DES PARAMÈTRES (Au début) ---
        mlflow.log_param("seuil_confiance_ocr", SEUIL_CONFIANCE_OCR)
        mlflow.log_param("nombre_abonnes", len(BASE_ABONNES))
        mlflow.log_param("modele", "YOLOv8_Custom")
        mlflow.log_param("ocr_engine", "EasyOCR")
        mlflow.log_param("fichier_video", nom_video)

        print(f" MLflow démarré ! Session liée à la vidéo {nom_video}")

        while cap.isOpened():
            ret, frame = cap.read()
            if not ret: break

            results = model.predict(source=frame, conf=0.6, verbose=False)
            
            x1_draw, y1_draw, x2_draw, y2_draw = 0, 0, 0, 0
            box_trouvee = False

            for r in results:
                for box in r.boxes:
                    box_trouvee = True
                    x1, y1, x2, y2 = map(int, box.xyxy[0])
                    x1_draw, y1_draw, x2_draw, y2_draw = x1, y1, x2, y2
                    
                    y1_safe, y2_safe = max(0, y1), min(frame.shape[0], y2)
                    x1_safe, x2_safe = max(0, x1), min(frame.shape[1], x2)
                    plaque_crop = frame[y1_safe:y2_safe, x1_safe:x2_safe]
                    
                    if plaque_crop.size == 0: continue
                    
                    plaque_prete = ameliorer_plaque(plaque_crop)
                    result_ocr = reader.readtext(plaque_prete)
                    
                    for (_, text, prob) in result_ocr:
                        t_clean = re.sub(r'[^A-Z0-9]', '', text.upper())
                        if len(t_clean) >= 4 and any(c.isdigit() for c in t_clean):
                            if t_clean not in LISTE_NOIRE_PROVINCES:
                                if prob > SEUIL_CONFIANCE_OCR:
                                    memoire_plaques.append(t_clean)
                                    toutes_plaques_detectees.add(t_clean) # Historique global

            if memoire_plaques and box_trouvee:
                occurences = Counter(memoire_plaques)
                plaque_stable, nb = occurences.most_common(1)[0]
                couleur_box = (0, 255, 0) if plaque_stable in BASE_ABONNES else (0, 0, 255)
                
                cv2.rectangle(frame, (x1_draw, y1_draw), (x2_draw, y2_draw), couleur_box, 3)
                cv2.putText(frame, plaque_stable, (x1_draw, max(0, y1_draw - 10)), 
                            cv2.FONT_HERSHEY_SIMPLEX, 0.8, couleur_box, 2)

            if len(memoire_plaques) > 20:
                memoire_plaques.pop(0)

            if memoire_plaques:
                occurences = Counter(memoire_plaques)
                plaque_stable, nb = occurences.most_common(1)[0]

                if nb > 1:
                    status = "AUTORISE" if plaque_stable in BASE_ABONNES else "NON AUTORISE"
                    couleur_statut = (0, 255, 0) if status == "AUTORISE" else (0, 0, 255)
                    
                    cv2.rectangle(frame, (10, 10), (450, 80), (0, 0, 0), -1)
                    cv2.putText(frame, f"PLAQUE: {plaque_stable}", (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
                    cv2.putText(frame, f"STATUT: {status}", (20, 70), cv2.FONT_HERSHEY_SIMPLEX, 0.8, couleur_statut, 2)
                    
                    if plaque_stable != derniere_plaque_affichee:
                        print(f"[{status}] Véhicule : {plaque_stable}")
                        mlflow.log_metric("confiance_vote", nb/20)
                        derniere_plaque_affichee = plaque_stable

            if not mode_headless:
                frame_redimensionne = cv2.resize(frame, (int(frame.shape[1] * 0.8), int(frame.shape[0] * 0.8)))
                cv2.imshow("Controle Parking La Cite - IA", frame_redimensionne)
                if cv2.waitKey(1) & 0xFF == ord('q'): break

        # --- ENREGISTREMENT DES MÉTRIQUES DE FIN (Résumé global) ---
        mlflow.log_metric("total_plaques_uniques_detectees", len(toutes_plaques_detectees))
        
        # Enregistre si la dernière plaque vue était autorisée (1) ou non (0)
        statut_final = 1 if derniere_plaque_affichee in BASE_ABONNES else 0
        mlflow.log_metric("statut_derniere_plaque", statut_final)

    cap.release()
    if not mode_headless: cv2.destroyAllWindows()
    print(f" Analyse vidéo terminée. Données sauvegardées dans {chemin_mlruns}")

if __name__ == "__main__":
    analyser_video_complete("IMG_4318.MOV", mode_headless=False)
```
